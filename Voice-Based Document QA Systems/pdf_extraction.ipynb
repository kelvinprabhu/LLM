```python
import PyPDF2
def extract_text(pdf_path):
    with open(pdf_path, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        text = " ".join([page.extract_text() for page in reader.pages])
    return text
```

```python
english_text = extract_text("pdfs/atentionalluneed.pdf")
tamil_text = extract_text("pdfs/tamil_baratiyar.pdf")
french_text = extract_text("pdfs/fenchstory.pdf")
```

```python
import re

def clean_text(text, language):
    # Basic cleaning: remove extra whitespace and non-printable characters
    text = re.sub(r'\s+', ' ', text).strip()
    if language == 'english':
        # Remove non-ASCII characters for English
        text = re.sub(r'[^\x00-\x7F]+', '', text)
    elif language == 'tamil':
        # Remove non-Tamil unicode blocks (Tamil: U+0B80–U+0BFF)
        text = ''.join([c for c in text if '\u0B80' <= c <= '\u0BFF' or c.isspace()])
    elif language == 'french':
        # Allow French accented characters, remove other non-printable
        text = re.sub(r'[^\x00-\x7F\u00C0-\u017F]+', '', text)
    return text

# Clean and save as markdown
with open('english_text.md', 'w', encoding='utf-8') as f:
    f.write(clean_text(english_text, 'english'))

with open('tamil_text.md', 'w', encoding='utf-8') as f:
    f.write(clean_text(tamil_text, 'tamil'))

```
